# Visualization
Create the final charts for target distribution, department performance and the top three model-derived factors.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_excel(r"C:\Users\karth\Desktop\cds_iabac\data\raw\INX_Future_Inc_Employee_Performance_CDS_Project2_Data_V1.8.xls")
importance = pd.read_csv(r"C:\Users\karth\Desktop\cds_iabac\project summary\Analysis\feature_importance.csv",index_col=0).squeeze("columns")
```

```python
plt.figure(figsize=(7,5))
sns.countplot(data=df,x="PerformanceRating")
plt.title("Performance Rating Distribution"); plt.tight_layout(); plt.show()
```

```python
dept = df.groupby("EmpDepartment")["PerformanceRating"].mean().sort_values(ascending=False)
plt.figure(figsize=(9,5))
sns.barplot(x=dept.values,y=dept.index)
plt.xlabel("Average Performance Rating"); plt.ylabel("Department")
plt.title("Average Performance Rating by Department"); plt.tight_layout(); plt.show()
```

```python
top3 = importance.head(3).sort_values()
plt.figure(figsize=(8,4))
sns.barplot(x=top3.values,y=top3.index)
plt.xlabel("Mean Permutation Importance"); plt.ylabel("Factor")
plt.title("Top Three Predictive Factors"); plt.tight_layout(); plt.show()
```
