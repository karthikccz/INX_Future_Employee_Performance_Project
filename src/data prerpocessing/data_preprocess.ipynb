# Data Processing

The original data is preserved in `data/raw`. The canonical modeling dataset is written to `data/processed`.

### Processing decisions
- 1,200 employee records and 28 columns were supplied.
- `EmpNumber` is an identifier and is excluded from modeling.
- `PerformanceRating` is the target.
- Missing values and duplicates are checked.
- No imputation is needed if the source remains unchanged.
- Categorical variables are one-hot encoded.
- Numerical variables are standardized for the Logistic Regression pipeline.
- No Principal Component Analysis is used because the project requires interpretable business factors.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
```

```python
DATASET = r"C:\\Users\\karth\Desktop\\cds_iabac\\data\\raw\\INX_Future_Inc_Employee_Performance_CDS_Project2_Data_V1.8.xls"
df = pd.read_excel(DATASET)
print("dataset loaded succesfully")
print("shape of dataset", df.shape)
display(df.head())
```

```python
col_list = df.select_dtypes(include="object").columns.tolist()
col_list
```

```python
df_encoded = pd.get_dummies(df,columns=col_list, drop_first=True, dtype=int)
df_encoded.head()
```

```python
num_list = df.select_dtypes(include="int").columns.tolist()
num_list
```

```python
sc = StandardScaler()
df_encoded[num_list] = sc.fit_transform(df_encoded[num_list])
df_encoded.head()
```

```python
df_encoded.to_csv(r"C:\Users\karth\Desktop\cds_iabac\data\preprocessed\employee_performance_clean.csv", index=False)
print("Canonical processed dataset saved.")
```
